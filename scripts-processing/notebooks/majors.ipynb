# 1. Khởi tạo môi trường & Đọc dữ liệu

```python
# export
import re
import pandas as pd
from helpers.get_configs import get_config

RAW_DATA_PATH = get_config('dirs.yaml', ['raw_data', 'majors'], as_path=True)
CLEANED_DATA_PATH = get_config('dirs.yaml', ['cleaned_data', 'majors'], as_path=True)
SCRIPT_OUTPUT_PATH = get_config('dirs.yaml', ['unit_scripts', 'majors'], as_path=True)

# Đọc file CSV vào DataFrame
df = pd.read_parquet(RAW_DATA_PATH)
```

```python
# Khảo sát thông tin tổng quan của dữ liệu
df.info()
df.head()
```

# 2. Kiểm tra giá trị rỗng

```python
# Kiểm tra số lượng giá trị rỗng trên các cột
df.isna().sum()
```

**Kết luận**: Các cột bị thiếu dữ liệu đều thuộc về `major_description` là cột không quan trọng, nên không cần bổ sung / lược bỏ các hàng này.

# 3. Chuẩn hóa danh sách và chuỗi

```python
# export
def remove_newlines(val):
    if pd.isna(val):
        return val
    if isinstance(val, str):
        return re.sub(r'[\r\n]+', ' ', val).strip()
    return val

df = df.map(remove_newlines)

def normalize_score_codes(val):
    if pd.isna(val) or str(val).strip() == '':
        return val
    s = str(val).replace(',', ';')
    s = re.sub(r'\s+', '', s)
    return s

df['score_codes'] = df['score_codes'].apply(normalize_score_codes)
```

# 4. Lưu dữ liệu đã làm sạch

```python
# export
CLEANED_DATA_PATH.parent.mkdir(parents=True, exist_ok=True)
df.to_parquet(CLEANED_DATA_PATH, index=False)
```
