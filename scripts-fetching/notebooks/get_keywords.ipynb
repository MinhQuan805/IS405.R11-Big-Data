```python
from pytrends.request import TrendReq
import pandas as pd

pytrends = TrendReq(hl='en-US', tz=360)
```

```python
pd.DataFrame(pytrends.suggestions("Trường Đại học Công nghệ thông tin")) 
```

```python
kw_list = ["/m/03ytc"]
pytrends.build_payload(kw_list, timeframe='today 5-y', geo='VI')
pytrends.interest_over_time()

# Các định dạng timeframe
# Mặc định: 'today 5-y' (5 năm gần nhất)
# Toàn bộ: 'all'
# Ngày cụ thể: '2016-12-14 2017-01-25'
# Ngày giờ cụ thể (theo UTC): '2017-02-06T10 2017-02-12T07'
# Theo tháng: 'today 3-m' (chỉ hỗ trợ 1, 3, 12 tháng)
# Theo ngày: 'now 7-d' (chỉ hỗ trợ 1, 7 ngày)
# Theo giờ: 'now 1-H' (chỉ hỗ trợ 1, 4 giờ)
```
